```python
import matplotlib.pyplot as plt

grades = ["G1", "G2", "G3"]
averages = [df["G1"].mean(), df["G2"].mean(), df["G3"].mean()]

plt.bar(grades, averages)
plt.title("Average Grades: G1 vs G2 vs G3")
plt.xlabel("Grade")
plt.ylabel("Average Score")
plt.show()
```

```python
studytime_g3 = df.groupby("studytime")["G3"].mean()

print(studytime_g3)
```

```python
studytime_g3.plot(kind="bar")

plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time")
plt.ylabel("Average G3")
plt.xticks(rotation=0)
plt.show()
```

```python
correlation = df.corr(numeric_only=True)["G3"].sort_values(ascending=False)

print(correlation)
```

```python
g3_corr = correlation.drop("G3")

g3_corr.plot(kind="bar", figsize=(10, 6))

plt.title("Correlation of Numerical Features with Final Grade (G3)")
plt.xlabel("Features")
plt.ylabel("Correlation with G3")
plt.xticks(rotation=45)
plt.axhline(0)
plt.tight_layout()
plt.show()
```
